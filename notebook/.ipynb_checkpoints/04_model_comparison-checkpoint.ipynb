```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error, median_absolute_error
```

```python
df = pd.read_csv("cleaned_crmls_sold.csv")

train_df = df[df['is_test'] == 0].copy()
test_df = df[df['is_test'] == 1].copy()

feature_cols = [col for col in df.columns if col not in ['target_log_price', 'is_test']]

X_train = train_df[feature_cols]
y_train = train_df['target_log_price']

X_test = test_df[feature_cols]
y_test = test_df['target_log_price']

y_test_original = np.expm1(y_test)
```

```python
models = {
    "Baseline (Linear Regression)": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42, max_depth=10),
    "Random Forest Regressor": RandomForestRegressor(random_state=42, n_estimators=100, max_depth=15, n_jobs=-1)
}

results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    
    y_pred_log = model.predict(X_test)
    r2 = r2_score(y_test, y_pred_log)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred_log))
    
    y_pred_original = np.expm1(y_pred_log)
    mae_dollar = mean_absolute_error(y_test_original, y_pred_original)
    medae_dollar = median_absolute_error(y_test_original, y_pred_original)
    
    results.append({
        "Model": name,
        "Test R² Score": round(r2, 4),
        "RMSE (Log)": round(rmse, 4),
        "MAE ($)": f"${mae_dollar:,.2f}",
        "Median AE ($)": f"${medae_dollar:,.2f}"
    })

results_df = pd.DataFrame(results)
print("=" * 70)
print("📊 Week 5 Mpdel result (vs Baseline)")
print("=" * 70)
print(results_df.to_string(index=False))
```

```python
rf_model = models["Random Forest Regressor"]
importances = rf_model.feature_importances_

feature_imp = pd.DataFrame({
    'Feature': feature_cols,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

print(feature_imp.head(10).to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import geopandas as gpd
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error, median_absolute_error
```

```python
df_fe = df.copy()

# 1) 침실 대비 욕실 비율 (bed/bath ratio)
# (컬럼명이 대소문자 차이가 있을 수 있어 자동 찾기 적용)
bed_col = [c for c in df_fe.columns if 'BED' in c.upper() or 'BEDROOM' in c.upper()]
bath_col = [c for c in df_fe.columns if 'BATH' in c.upper() or 'BATHROOM' in c.upper()]

if bed_col and bath_col:
    df_fe['bed_bath_ratio'] = df_fe[bed_col[0]] / (df_fe[bath_col[0]] + 1)
    print("  -> 'bed_bath_ratio' 생성 완료!")

# 2) 주택 연식 (age of property in years)
year_col = [c for c in df_fe.columns if 'YEAR' in c.upper() or 'BUILT' in c.upper()]
if year_col:
    df_fe['property_age'] = 2026 - df_fe[year_col[0]]
    print("  -> 'property_age' 생성 완료!")

# 3) 방 하나당 평균 면적
sqft_col = [c for c in df_fe.columns if 'SQFT' in c.upper() or 'AREA' in c.upper() or 'LIVING' in c.upper()]
if sqft_col and bed_col:
    df_fe['sqft_per_bedroom'] = df_fe[sqft_col[0]] / (df_fe[bed_col[0]] + 1)
    print("  -> 'sqft_per_bedroom' 생성 완료!")


# =====================================================================
# [Step 2] 위치(Spatial) 컬럼 존재 여부 확인 및 Spatial Join 시도
# =====================================================================
lat_cols = [c for c in df_fe.columns if 'LAT' in c.upper()]
lon_cols = [c for c in df_fe.columns if 'LON' in c.upper() or 'LNG' in c.upper()]

if lat_cols and lon_cols:
    print(f"🗺️ 위치 컬럼 감지됨 ({lat_cols[0]}, {lon_cols[0]}). 학군 데이터 공간 결합 시도 중...")
    try:
        import geopandas as gpd
        geojson_url = "https://data.ca.gov/dataset/e2112e3e-324c-4a30-a249-14a51e626e2e/resource/7dfaf005-58eb-45db-93b1-7aff091b2172/download/ca-school-district-areas-2024-25.geojson"
        
        gdf_properties = gpd.GeoDataFrame(
            df_fe, 
            geometry=gpd.points_from_xy(df_fe[lon_cols[0]], df_fe[lat_cols[0]]),
            crs="EPSG:4326"
        )
        school_districts = gpd.read_file(geojson_url)
        if gdf_properties.crs != school_districts.crs:
            school_districts = school_districts.to_crs(gdf_properties.crs)
            
        joined_gdf = gpd.sjoin(gdf_properties, school_districts, how="left", predicate="intersects")
        school_col = [c for c in school_districts.columns if 'NAME' in c.upper() or 'DISTRICT' in c.upper()][0]
        
        df_fe = pd.DataFrame(joined_gdf.drop(columns=['geometry', 'index_right'], errors='ignore'))
        df_fe[school_col] = df_fe[school_col].fillna('Unknown')
        df_fe = pd.get_dummies(df_fe, columns=[school_col], drop_first=True)
        print("  -> 학군 데이터 성공적으로 연결 완료!")
    except Exception as e:
        print(f"  ⚠️ Spatial Join 건너뜀: {e}")
else:
    print("ℹ️ 원본 위도/경도 컬럼이 전처리 단계에서 정규화/제외되었으므로 공간 조인을 건너뛰고 생성된 파생 변수만으로 진행합니다.")
```

```python
geojson_url = "https://data.ca.gov/dataset/e2112e3e-324c-4a30-a249-14a51e626e2e/resource/7dfaf005-58eb-45db-93b1-7aff091b2172/download/ca-school-district-areas-2024-25.geojson"

try:
    # 1) GeoDataFrame 변환 (위도, 경도 -> Point)
    gdf_properties = gpd.GeoDataFrame(
        df_fe, 
        geometry=gpd.points_from_xy(df_fe['Longitude'], df_fe['Latitude']),
        crs="EPSG:4326"
    )

    # 2) URL 주소에서 GeoJSON 파일 direct 로드
    school_districts = gpd.read_file(geojson_url)

    # 3) 좌표계(CRS) 맞추기
    if gdf_properties.crs != school_districts.crs:
        school_districts = school_districts.to_crs(gdf_properties.crs)

    # 4) Spatial Join 실행 (집 위치 좌표가 어떤 학군 구역에 속하는지 연결)
    joined_gdf = gpd.sjoin(gdf_properties, school_districts, how="left", predicate="intersects")

    # 5) 학군 컬럼 감지 및 원핫 인코딩
    school_col = [c for c in school_districts.columns if 'NAME' in c.upper() or 'DISTRICT' in c.upper() or 'CDCode' in c][0]
    print(f"  -> 감지된 학군 정보 컬럼: '{school_col}'")

    df_fe = pd.DataFrame(joined_gdf.drop(columns=['geometry', 'index_right'], errors='ignore'))
    df_fe[school_col] = df_fe[school_col].fillna('Unknown')

    # 학군 범주원핫 인코딩
    df_fe = pd.get_dummies(df_fe, columns=[school_col], drop_first=True)
    print("  -> 학군 데이터 성공적으로 연결 완료!")

except Exception as e:
    print(f"  ⚠️ URL로 읽어오는 중 에러 발생: {e}")
    print("  -> 학군 공간 데이터를 제외하고 기본 파생 변수만으로 학습을 진행합니다.")
```
